# 05 - Customer Segmentation 👥

**Goal:** group customers into meaningful segments.

**Segment by:** country, customer type, length of stay, party size

**Example segments to look for:**
- **Business travelers:** short stay, weekday-heavy
- **Vacation travelers:** longer stay, weekend-heavy

```python
import sys
sys.path.append('..')

import pandas as pd

from src.config import CLEAN_DATA_PATH
from src.feature_engineering import add_stay_features
from src.analysis import customer_type_profile, business_vs_vacation_segments

df = pd.read_csv(CLEAN_DATA_PATH)
df = add_stay_features(df)
```

## Weekday vs weekend stay ratio

```python
weekend_share = (df['stays_in_weekend_nights'] / df['total_stay'].replace(0, 1))
print(f'Median weekend-night share across all stays: {weekend_share.median():.1%}')
```

## Segment: Business vs Vacation travelers

```python
segment_profile = business_vs_vacation_segments(df)
segment_profile
```

## Segment by customer type

```python
ct_profile = customer_type_profile(df)
ct_profile
```

## Key insight

```python
biz = segment_profile.loc['Business traveler']
vac = segment_profile.loc['Vacation traveler']
print(f"- Business-traveler-style bookings ({biz['share_of_bookings']:.1%} of the total) cancel at {biz['cancellation_rate']:.1%}, vs {vac['cancellation_rate']:.1%} for vacation-style bookings.")
print(f"- Vacation-style bookings run longer ({vac['avg_stay']:.1f} nights) and pay a higher ADR ({vac['avg_adr']:.0f}) on average than business-style bookings ({biz['avg_stay']:.1f} nights, {biz['avg_adr']:.0f} ADR).")
```
