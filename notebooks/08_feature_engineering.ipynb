# 08 - Feature Engineering

**Goal:** build the features needed for modeling and save a model-ready dataset.

**Input:** `data/processed/hotel_bookings_clean.csv`
**Output:** `data/processed/hotel_bookings_features.csv`

```python
import sys
sys.path.append('..')

import pandas as pd

from src.config import CLEAN_DATA_PATH, FEATURES_DATA_PATH
from src.feature_engineering import engineer_features
from src.data_cleaning import drop_leakage_columns

df = pd.read_csv(CLEAN_DATA_PATH)
df.shape
```

## Run the feature engineering pipeline

Adds: `total_members`, `total_guests`, `has_family`, `total_stay`, `room_changed`, `long_lead_time`, `arrival_quarter`, `total_previous_bookings`, `prev_cancel_ratio`, `revenue`, and encodes `deposit_type`.

```python
df = engineer_features(df)
df.shape
```

```python
df[['total_members', 'total_guests', 'has_family', 'total_stay',
    'room_changed', 'long_lead_time', 'arrival_quarter',
    'total_previous_bookings', 'prev_cancel_ratio', 'revenue']].describe(include='all')
```

## Drop leakage columns

`reservation_status` and `reservation_status_date` are only known **after** the cancellation outcome, so they must not be used as predictors.

```python
df = drop_leakage_columns(df)
```

## Save feature-engineered dataset

```python
df.to_csv(FEATURES_DATA_PATH, index=False)
print(f'Saved {len(df):,} rows, {df.shape[1]} columns to {FEATURES_DATA_PATH}')
```
